# 📊 Exploratory Data Analysis (EDA)

In this phase, we explore the dataset to understand employee characteristics, identify patterns, and discover factors that may influence employee attrition.

## 👥 Employee Attrition Distribution

Analyze the distribution of employees who stayed versus those who left the company.

````python
df_clean["Attrition"].value_counts()
````

## 📊 Attrition Bar Chart

Visualize the number of employees who stayed and those who left the company.

````python
# ==========================================
# Attrition Bar Chart
# ==========================================

attrition_counts = df_clean["Attrition"].value_counts()

plt.figure(figsize=(8, 5))

bars = plt.bar(
    attrition_counts.index,
    attrition_counts.values,
    color=["#4CAF50", "#E74C3C"],
    edgecolor="black",
    linewidth=1.2
)

plt.title(
    "Employee Attrition Distribution",
    fontsize=16,
    fontweight="bold",
    pad=15
)

plt.xlabel(
    "Attrition",
    fontsize=12
)

plt.ylabel(
    "Number of Employees",
    fontsize=12
)

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.3
)

# Add values on top of bars
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 10,
        f"{int(height)}",
        ha="center",
        fontsize=11,
        fontweight="bold"
    )

plt.tight_layout()

plt.show()
````

## 👨‍💼👩‍💼 Attrition by Gender

Analyze employee attrition based on gender.

````python
# ==========================================
# Attrition by Gender
# ==========================================

plt.figure(figsize=(8, 5))

ax = sns.countplot(
    data=df_clean,
    x="Gender",
    hue="Attrition",
    palette="Set2"
)

plt.title(
    "Employee Attrition by Gender",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Gender", fontsize=12)
plt.ylabel("Number of Employees", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.3)

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()

plt.show()
````

### 📈 Attrition Rate by Gender

Calculate the attrition rate for each gender to compare proportions rather than raw counts.

````python
# ==========================================
# Attrition Rate by Gender
# ==========================================

gender_attrition = (
    pd.crosstab(
        df_clean["Gender"],
        df_clean["Attrition"],
        normalize="index"
    ) * 100
)

gender_attrition.round(2)
````

### 📌 Insight

- Male employees have a slightly higher attrition rate (17.01%) than female employees (14.80%).
- Although more male employees left the company in absolute numbers, comparing percentages provides a more accurate assessment because the workforce contains more male employees.
- Gender alone does not appear to be a strong predictor of employee attrition, as the difference between the two groups is relatively small.

## 🏢 Attrition by Department

Analyze employee attrition across different departments.

````python
# ==========================================
# Attrition by Department
# ==========================================

plt.figure(figsize=(9, 5))

ax = sns.countplot(
    data=df_clean,
    x="Department",
    hue="Attrition",
    palette="Set2"
)

plt.title(
    "Employee Attrition by Department",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Department", fontsize=12)
plt.ylabel("Number of Employees", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.3)

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()

plt.show()
````

### 📈 Attrition Rate by Department

Calculate the employee attrition rate for each department.

````python
# ==========================================
# Attrition Rate by Department
# ==========================================

department_attrition = (
    pd.crosstab(
        df_clean["Department"],
        df_clean["Attrition"],
        normalize="index"
    ) * 100
)

department_attrition.round(2)
````

### 📌 Insight

- The **Sales** department has the highest attrition rate (**20.63%**), indicating a higher proportion of employees leaving compared to other departments.
- **Human Resources** follows with an attrition rate of **19.05%**.
- Although **Research & Development** had the highest number of employees leaving in absolute terms, it has the **lowest attrition rate (13.84%)**, reflecting its larger workforce.
- Comparing percentages rather than raw counts provides a more accurate assessment of employee turnover across departments.

## ⏰ Attrition by OverTime

Analyze employee attrition based on overtime status.

````python
# ==========================================
# Attrition by OverTime
# ==========================================

plt.figure(figsize=(9, 5))

ax = sns.countplot(
    data=df_clean,
    x="OverTime",
    hue="Attrition",
    palette={
        "No": "#2E8B57",
        "Yes": "#DC143C"
    }
)

plt.title(
    "Employee Attrition by OverTime",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("OverTime", fontsize=12)
plt.ylabel("Number of Employees", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.3)

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()

plt.show()
````

### 📈 Attrition Rate by OverTime

Calculate the attrition rate for employees who work overtime versus those who do not.

````python
# ==========================================
# Attrition Rate by OverTime
# ==========================================

overtime_attrition = (
    pd.crosstab(
        df_clean["OverTime"],
        df_clean["Attrition"],
        normalize="index"
    ) * 100
)

overtime_attrition.round(2)
````

### 📌 Insight

- Employees who work overtime have a significantly higher attrition rate (**30.53%**) compared to employees who do not work overtime (**10.44%**).
- This suggests that overtime may be a major factor contributing to employee turnover.
- Reducing excessive overtime or improving workload management could help lower attrition rates.

## 💰 Monthly Income vs Attrition

Compare the distribution of monthly income between employees who stayed and those who left the company.

````python
# ==========================================
# Monthly Income vs Attrition
# ==========================================

plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df_clean,
    x="Attrition",
    y="MonthlyIncome",
    hue="Attrition",
    palette={
        "No": "#2E8B57",
        "Yes": "#DC143C"
    },
    legend=False
)

plt.title(
    "Monthly Income by Attrition",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Attrition", fontsize=12)
plt.ylabel("Monthly Income", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()

plt.show()
````

### 📌 Insight

- Employees who left the company generally had lower monthly incomes than those who stayed.
- The median monthly income is noticeably higher among employees who remained with the company.
- This suggests that lower income may be associated with a higher likelihood of employee attrition.
- Both groups contain a few high-income outliers, indicating the presence of senior or highly paid employees.

## 📅 Years at Company vs Attrition

Compare the distribution of years at the company between employees who stayed and those who left.

````python
# ==========================================
# Years at Company vs Attrition
# ==========================================

plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df_clean,
    x="Attrition",
    y="YearsAtCompany",
    hue="Attrition",
    palette={
        "No": "#2E8B57",
        "Yes": "#DC143C"
    },
    legend=False
)

plt.title(
    "Years at Company by Attrition",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Attrition", fontsize=12)
plt.ylabel("Years at Company", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()

plt.show()
````

## 🔥 Correlation Analysis

Analyze the relationships between numerical features using a correlation matrix.

## 📊 Correlation Heatmap

Visualize the correlation between numerical variables.

````python
# ==========================================
# Correlation Heatmap
# ==========================================

# Select only numerical columns
numeric_df = df_clean.select_dtypes(include="number")

# Calculate correlation matrix
correlation_matrix = numeric_df.corr()

# Plot heatmap
plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="RdYlBu",
    linewidths=0.5
)

plt.title(
    "Correlation Heatmap",
    fontsize=18,
    fontweight="bold"
)

plt.tight_layout()

plt.show()
````

## 🔝 Top Correlated Features

Identify the strongest correlations between numerical features.

````python
# ==========================================
# Top Correlated Features
# ==========================================

corr_matrix = numeric_df.corr()

corr_pairs = (
    corr_matrix
    .where(~np.eye(corr_matrix.shape[0], dtype=bool))
    .stack()
    .sort_values(ascending=False)
)

top_corr = corr_pairs.drop_duplicates().head(10)

top_corr
````

### 📌 Insight

- The strongest positive correlation was observed between **Job Level** and **Monthly Income (0.95)**, indicating that employees in higher job levels generally earn higher salaries.
- **Years at Company** is strongly associated with **Years with Current Manager (0.77)** and **Years in Current Role (0.76)**, reflecting expected career progression.
- **Total Working Years** has a strong positive relationship with **Monthly Income (0.77)**, suggesting that employees with more experience tend to receive higher salaries.
- Most remaining feature pairs show weak correlations, indicating limited linear relationships among them.

## 📊 Top 10 Correlated Features

Display the strongest positive correlations between numerical features.

````python
# ==========================================
# Top 10 Correlated Features
# ==========================================

# Create correlation matrix
corr_matrix = numeric_df.corr()

# Keep only the upper triangle
corr_pairs = (
    corr_matrix.where(
        np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
    )
    .stack()
    .sort_values(ascending=False)
)

# Select top 10 correlations
top_corr = corr_pairs.head(10)

# Convert to DataFrame
top_corr_df = top_corr.reset_index()
top_corr_df.columns = [
    "Feature 1",
    "Feature 2",
    "Correlation"
]

top_corr_df
````

## 📈 Top 10 Correlations Visualization

Visualize the strongest correlations between numerical features.

````python
# ==========================================
# Top 10 Correlations Visualization
# ==========================================

plt.figure(figsize=(11, 6))

labels = (
    top_corr_df["Feature 1"]
    + " ↔ "
    + top_corr_df["Feature 2"]
)

ax = sns.barplot(
    data=top_corr_df,
    x="Correlation",
    y=labels,
    palette="viridis",
    hue=labels,
)

plt.title(
    "Top 10 Strongest Feature Correlations",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Correlation Coefficient")
plt.ylabel("")

plt.grid(axis="x", linestyle="--", alpha=0.3)

# Add values on bars
for container in ax.containers:
    ax.bar_label(container, fmt="%.2f")

plt.tight_layout()

plt.show()
````

### 📌 Insight

- **Job Level** and **Monthly Income** have the strongest positive correlation (**0.95**), indicating that employees in higher job levels tend to receive higher salaries.
- **Years at Company** is strongly associated with both **Years with Current Manager** and **Years in Current Role**, reflecting expected career progression within the organization.
- **Total Working Years** also shows a strong positive relationship with **Monthly Income**, suggesting that greater work experience is generally associated with higher earnings.
- Most other numerical features exhibit weak correlations, indicating that they are relatively independent of one another.

# 📊 KPI Analysis

Calculate key performance indicators (KPIs) to provide a high-level overview of the workforce.

## 📌 Workforce Overview

````python
# ==========================================
# Workforce KPIs
# ==========================================

# Total employees
total_employees = len(df_clean)

# Employees who left
employees_left = (df_clean["Attrition"] == "Yes").sum()

# Employees who stayed
employees_stayed = (df_clean["Attrition"] == "No").sum()

# Attrition rate
attrition_rate = (
    employees_left / total_employees
) * 100

print(f"Total Employees : {total_employees}")
print(f"Employees Left  : {employees_left}")
print(f"Employees Stayed: {employees_stayed}")
print(f"Attrition Rate  : {attrition_rate:.2f}%")
````

### 📋 Workforce KPIs

````python
# ==========================================
# Workforce KPI Table
# ==========================================

kpi_df = pd.DataFrame({
    "KPI": [
        "Total Employees",
        "Employees Left",
        "Employees Stayed",
        "Attrition Rate"
    ],
    "Value": [
        total_employees,
        employees_left,
        employees_stayed,
        f"{attrition_rate:.2f}%"
    ]
})

kpi_df
````

````python
# ==========================================
# Employee Statistics KPIs
# ==========================================

average_age = df_clean["Age"].mean()

average_income = df_clean["MonthlyIncome"].mean()

average_years = df_clean["YearsAtCompany"].mean()

average_job_satisfaction = df_clean["JobSatisfaction"].mean()

average_worklife = df_clean["WorkLifeBalance"].mean()
````

````python
additional_kpis = pd.DataFrame({
    "KPI": [
        "Average Age",
        "Average Monthly Income",
        "Average Years at Company",
        "Average Job Satisfaction",
        "Average Work-Life Balance"
    ],
    "Value": [
        f"{average_age:.1f}",
        f"${average_income:,.0f}",
        f"{average_years:.1f}",
        f"{average_job_satisfaction:.2f}",
        f"{average_worklife:.2f}"
    ]
})

additional_kpis
````
